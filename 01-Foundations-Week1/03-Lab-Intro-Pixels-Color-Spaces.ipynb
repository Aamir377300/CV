## What is Computer Vision?
Computer Vision is the field of AI that enables machines to interpret and understand visual information.
Applications include:
- Self-driving cars
- Medical imaging
- Surveillance
- Augmented Reality

```python
import cv2
import matplotlib.pyplot as plt

# Load image
img = cv2.imread('sample.jpg')  # replace with your image file
img_rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

print("Image shape:", img.shape)  # (height, width, channels)

# Display image
plt.imshow(img_rgb)
plt.title("Original Image")
plt.axis('off')
plt.show()
```

## Images as Data
- Digital images are grids of pixels.
- Each pixel has intensity values (RGB channels).
- Resolution defines detail.

```python
# Inspect pixel values
print("Pixel at (50,50):", img[50,50])  # BGR values
```

## Color Spaces
- RGB: Red, Green, Blue
- HSV: Hue, Saturation, Value
Converting between color spaces helps in tasks like color detection.

```python
hsv = cv2.cvtColor(img, cv2.COLOR_BGR2HSV)

plt.imshow(cv2.cvtColor(hsv, cv2.COLOR_HSV2RGB))
plt.title("HSV Color Space")
plt.axis('off')
plt.show()
```

## Transformations Overview
- Translation: shifting image
- Scaling: resizing
- Rotation: turning image around a pivot
- Perspective: warping viewpoint

## Filtering & Convolutions
- Convolution = sliding a kernel over an image.
- Filters detect edges, textures, and patterns.
- This is the foundation of CNNs.

```python
import numpy as np

kernel = np.array([[0,-1,0],[-1,5,-1],[0,-1,0]])
sharp = cv2.filter2D(img, -1, kernel)

plt.imshow(cv2.cvtColor(sharp, cv2.COLOR_BGR2RGB))
plt.title("Sharpened Image")
plt.axis('off')
plt.show()
```

```python
edges = cv2.Canny(img, 100, 200)

plt.imshow(edges, cmap='gray')
plt.title("Edge Detection")
plt.axis('off')
plt.show()
```

## Mini Project: Image Viewer
Build a simple viewer that:
- Loads an image
- Converts to grayscale
- Applies edge detection
Try toggling between modes!

```python
def image_viewer(img_path, mode="original"):
    img = cv2.imread(img_path)
    if mode == "gray":
        img = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)
    elif mode == "edges":
        img = cv2.Canny(img, 100, 200)
    plt.imshow(img, cmap='gray' if mode!="original" else None)
    plt.title(f"Mode: {mode}")
    plt.axis('off')
    plt.show()

# Try different modes
image_viewer('sample.jpg', mode="original")
image_viewer('sample.jpg', mode="gray")
image_viewer('sample.jpg', mode="edges")
```
