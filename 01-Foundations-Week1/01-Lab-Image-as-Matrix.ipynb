# Image Representation as a Matrix
This notebook demonstrates how a small grayscale image can be represented as a matrix of numbers and visualized with Matplotlib.

```python
import numpy as np
import matplotlib.pyplot as plt
```

## Create a Tiny Grayscale Image
We’ll define a 5×5 matrix with values ranging from 0 (black) to 255 (white).

```python
# Create a tiny 5x5 grayscale image (values 0–255)
img_matrix = np.array([
    [0, 50, 100, 150, 200],
    [25, 75, 125, 175, 225],
    [50, 100, 150, 200, 250],
    [75, 125, 175, 225, 255],
    [100, 150, 200, 250, 255]
], dtype=np.uint8)

img_matrix
```

## Visualize the Image and Pixel Values
We’ll plot the image on the left and overlay the pixel intensity values on the right.

```python
# Plot the image and its matrix values
fig, ax = plt.subplots(1, 2, figsize=(8,4))

# Show image
ax[0].imshow(img_matrix, cmap='gray', vmin=0, vmax=255)
ax[0].set_title("Grayscale Image")
ax[0].axis('off')

# Show matrix values
ax[1].imshow(img_matrix, cmap='gray', vmin=0, vmax=255)
for i in range(img_matrix.shape[0]):
    for j in range(img_matrix.shape[1]):
        ax[1].text(j, i, img_matrix[i, j], ha='center', va='center', color='red')
ax[1].set_title("Pixel Intensity Matrix")
ax[1].axis('off')

plt.show()
```
